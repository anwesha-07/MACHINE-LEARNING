```python
n=data.shape[0]
print(n)
```

```python
x=data['X']
```

```python
y=data['Y']
```

```python
x.shape
```

```python
beta_1=(1/n*np.sum(x)*np.sum(y)-np.sum(x*y))/(1/n*(np.sum(x)**2)-np.sum(x*x))
print(beta_1)
```

```python
beta_0=1/n*np.sum(y)-beta_1/n*np.sum(x)
print(beta_0)
```

```python
plt.scatter(x,y)
x1=np.linspace(0,1800,500)
y1=beta_0+beta_1*x1
plt.plot(x1,y1,'-')
plt.show()
```
